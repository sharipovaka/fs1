# Формула Тейлора: численный эксперимент

Ноутбук проверяет теоретическую оценку остаточного члена ряда Маклорена
и показывает, как быстро растёт точность приближения с числом слагаемых.

Запускается в Google Colab без установки дополнительных пакетов.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial

plt.rcParams.update({"figure.figsize": (9, 5), "font.size": 12, "grid.alpha": 0.3})
```

## 1. Частичная сумма ряда Маклорена для sin(x)

$$\sin x = \sum_{k=0}^{n-1} \frac{(-1)^k x^{2k+1}}{(2k+1)!} + R_n(x)$$

```python
def taylor_sin(x, n_terms=5):
    """Частичная сумма ряда Маклорена для sin(x) из n_terms слагаемых."""
    x = np.asarray(x, dtype=float)
    result = np.zeros_like(x)
    for k in range(n_terms):
        result += (-1) ** k * x ** (2 * k + 1) / factorial(2 * k + 1)
    return result


for x in (0.1, 0.5, 1.0, 2.0, 3.0):
    approx = float(taylor_sin(x, 5))
    exact = np.sin(x)
    print(f"x={x:4.1f}  ряд={approx:14.10f}  sin={exact:14.10f}  ошибка={abs(approx - exact):.2e}")
```

## 2. Сравнение приближений разного порядка

```python
x = np.linspace(-2 * np.pi, 2 * np.pi, 600)

plt.plot(x, np.sin(x), "k", lw=2.5, label="sin x")
for n in (1, 2, 3, 5):
    plt.plot(x, taylor_sin(x, n), lw=1.6, label=f"{n} слагаемых")

plt.ylim(-2, 2)
plt.grid(True)
plt.legend()
plt.title("Приближение sin x частичными суммами ряда Маклорена")
plt.show()
```

## 3. Проверка оценки остаточного члена

Форма Лагранжа даёт оценку $|R_n(x)| \le \dfrac{|x|^{2n+1}}{(2n+1)!}$.
Сравним фактическую ошибку с этой границей.

```python
x0 = 1.5
print(f"{'n':>3} {'фактическая ошибка':>22} {'оценка Лагранжа':>20}")
for n in range(1, 9):
    actual = abs(float(taylor_sin(x0, n)) - np.sin(x0))
    bound = abs(x0) ** (2 * n + 1) / factorial(2 * n + 1)
    print(f"{n:>3} {actual:>22.3e} {bound:>20.3e}")
```
