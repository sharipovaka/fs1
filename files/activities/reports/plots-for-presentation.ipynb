# Графики для доклада: как оформить, чтобы читалось из зала

**Лаборатория математики ФН1**

Ноутбук показывает, как готовить иллюстрации для презентации: крупный шрифт,
подписанные оси, экспорт в векторный формат.
Открывается в Google Colab, дополнительная установка пакетов не нужна.

```python
import numpy as np
import matplotlib.pyplot as plt

# Единый стиль для всех графиков доклада.
# Шрифт крупный: то, что мелко на экране ноутбука, из зала не видно вовсе.
PRESENTATION_STYLE = {
    "figure.figsize": (8, 5),
    "figure.dpi": 120,
    "font.size": 16,
    "axes.titlesize": 18,
    "axes.labelsize": 16,
    "lines.linewidth": 2.5,
    "grid.alpha": 0.3,
    "legend.frameon": False,
    "savefig.bbox": "tight",
}
plt.rcParams.update(PRESENTATION_STYLE)

# Фирменная палитра: тёмно-синий и золотой
NAVY, ACCENT, GREY = "#123a5f", "#e0a92e", "#8a97a3"
```

## 1. График сходимости в логарифмическом масштабе

Стандартная иллюстрация для доклада по численным методам: рядом с расчётными
точками показывается опорная прямая ожидаемого порядка.

```python
h = np.array([0.1, 0.05, 0.025, 0.0125, 0.00625])
error = np.array([2.1e-3, 5.3e-4, 1.3e-4, 3.3e-5, 8.2e-6])

fig, ax = plt.subplots()
ax.loglog(h, error, "o-", color=NAVY, label="расчёт")
ax.loglog(h, error[0] * (h / h[0]) ** 2, "--", color=GREY, label=r"опорная $O(h^2)$")

ax.set_xlabel("шаг сетки $h$")
ax.set_ylabel("погрешность $\\max|u - u_h|$")
ax.set_title("Сходимость разностной схемы")
ax.grid(True, which="both")
ax.legend()

observed = np.log2(error[:-1] / error[1:])
print("наблюдаемый порядок:", np.round(observed, 2))
plt.show()
```

## 2. Сравнение методов на одном поле

Не больше трёх-четырёх кривых на график: пятая уже не различается из зала.

```python
iterations = np.arange(1, 26)

fig, ax = plt.subplots()
ax.semilogy(iterations, 0.9**iterations, color=NAVY, label="простая итерация")
ax.semilogy(iterations, 0.6**iterations, color=ACCENT, label="сопряжённые градиенты")
ax.semilogy(iterations, 0.3**iterations, color=GREY, label="с предобуславливателем")

ax.set_xlabel("номер итерации")
ax.set_ylabel("норма невязки")
ax.set_title("Скорость сходимости итерационных методов")
ax.grid(True, which="both")
ax.legend()
plt.show()
```

## 3. Экспорт в презентацию

Для Beamer и PowerPoint сохраняйте векторный PDF: он не размывается
