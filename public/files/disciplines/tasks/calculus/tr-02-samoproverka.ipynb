Укажите ниже параметры своего варианта из файла `tr-02-variants.csv`.

```python
import sympy as sp

sp.init_printing()

# --- параметры варианта ---
A_PARAM = 1   # задача 1
K_PARAM = 2   # задача 3
B_PARAM = 1   # задача 4
C_PARAM = 1   # задача 5

x, y = sp.symbols("x y", real=True)
```

## Задача 1. Неявное дифференцирование

$x^3 + y^3 - 3axy = 0$

```python
y_func = sp.Function("y")(x)
equation = x**3 + y_func**3 - 3 * A_PARAM * x * y_func

first = sp.solve(sp.Eq(sp.diff(equation, x), 0), sp.diff(y_func, x))[0]
print("y' =")
sp.pprint(sp.simplify(first))
```

## Задача 2. Предел по правилу Лопиталя

```python
limit_value = sp.limit((sp.tan(x) - x) / (x - sp.sin(x)), x, 0)
print("предел =", limit_value)
```

## Задача 3. Разложение по формуле Маклорена

```python
series = sp.series(sp.log(1 + K_PARAM * x), x, 0, 5)
sp.pprint(series)
```

## Задача 4. Исследование функции

```python
f = x**3 / (x**2 - B_PARAM)

print("Вертикальные асимптоты:", sp.solve(sp.denom(sp.together(f)), x))
print("Наклонная асимптота:   ", sp.limit(f / x, x, sp.oo), "* x +",
      sp.limit(f - x * sp.limit(f / x, x, sp.oo), x, sp.oo))
print("Критические точки:     ", sp.solve(sp.diff(f, x), x))
print("Точки перегиба:        ", sp.solve(sp.diff(f, x, 2), x))
```

```python
sp.plot(f, (x, -5, 5), ylim=(-15, 15), title="График исследуемой функции")
```

## Задача 5. Наибольшее и наименьшее значения на отрезке

```python
g = x * sp.exp(-C_PARAM * x)
critical = [p for p in sp.solve(sp.diff(g, x), x) if p.is_real and 0 <= p <= 3]
points = [sp.Integer(0), sp.Integer(3), *critical]
values = {p: sp.nsimplify(g.subs(x, p)) for p in points}

for point, value in values.items():
    print(f"f({point}) = {value} ≈ {float(value):.6f}")
```

## Задача 6. Оптимальное сечение балки

При $b^2 + h^2 = 4R^2$ максимизируем $W = bh^2/6$. Проверьте аналитический ответ.

```python
R, b = sp.symbols("R b", positive=True)
W = b * (4 * R**2 - b**2) / 6
b_opt = sp.solve(sp.diff(W, b), b)[0]
print("b =", sp.simplify(b_opt))
print("h =", sp.simplify(sp.sqrt(4 * R**2 - b_opt**2)))
```
